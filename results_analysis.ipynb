```python
import pandas as pd

results_path = 'emo_search_aggregate_results.csv'

results_df = pd.read_csv(results_path)
```

## Emotion frequency overall

The most frequent emotions were interest, struggle, and surprise. With my script, I did not find many expressions of happiness or annoyance.

```python
results_df[['Emotion' ,'Word']].groupby(['Emotion']).count().sort_values(by='Word', ascending=False)
```

## Word frequency overall

The course participants seemed to find many things interesting. This might point to some limitations in vocabulary regarding expressing interest and fascination in different aspects. Also, 'interesting' can be a generalized word used when one is not sure, how else to describe a certain result in the moment. The second most common word is challenges: it's likely due to the nature of the course; the students often discuss 'challenges' of visualisations.

From my manual validation process, I can say the word 'value' was the biggest offender when it comes to the false positives: I decided to keep it in the search dictionary to hopefully catch an expression like 'In this course, I valued X and Y.', but most of the time it was just variable values. Here, the AI method would've been much more effective!

```python
results_df[['Emotion' ,'Word']].groupby(['Word']).count().sort_values(by='Emotion', ascending=False)
```
